## IF statement 

Now that we have some context on variables and their types, let's dive into what makes the logic of python possible. IF is generally the first statement people learn and it does exactly what its name suggests: it checks a condition. 

Example: In Zurich, you qualify for a ZVV discount IF you're under 25 year old. Lets say we wanted to write a program that checks this. That would look like this:

```python
student_name = "Alex"
age = 22  # You can change this number to test different ages


x = 1

x==1

# Checking the age limit for the ZVV youth discount (under 25)
if age < 25:
    print(f"Hi {student_name}! You qualify for the ZVV Youth discount.")
    print("Your monthly subscription price is: CHF 64")
else:
    print(f"Hi {student_name}! You are 25 or older.")
    print("You must pay the standard adult monthly subscription price.")
```

In the code above, the statement ELSE is a short way of saying: "In any other case, do this". 

So if the age of a student is under 25 the program will print what can be found indented under the if statement. 

If the student's age is exactly 25 or over 25 (>25) then the program would print what is indented under else statement.

### ELIF and Number checks:

1. Equals: a == b
2. Not Equals: a != b
3. Less than: a < b
4. Less than or equal to: a <= b
5. Greater than: a > b
6. Greater than or equal to: a >= b

But hey, what if the condition is a bit more complicated that this? What if we had to determine if a person is a child, teenager or an adult based on their age? 

There is a more elegant way of writing this by using the ELIF statement like so:

```python
age = 18  # You can change this number to test different ages

# Checking the age limit for the ZVV youth discount (under 25)
if age < 13:
    print("This is a child")
elif (age>=13) and (age<18):
    print(f"This is a teenager of age {age}")

else:
    print("This is an adult")
```

### Nested if statement

What if you have to check multiple things IN A ROW? What if they cannot be checked simultaneously? Python has the answer too :). 

Just write multiple IFs one after each other, this is called a nested if.

```python
x = 19

if x > 10:
    print("Above ten,")
    if x > 20:
        print("and also above 20!")
    elif x == 19:
        print("this is nineteen")
```

## WHILE loop

Loops are slightly different than statements. As their name suggests they loop / repeat. The while loop repeats unti some sort of a condition is broken. Let me show you:

```python
i = 1
while i < 6:
    print(i)
    i += 1
```

Here, the condition is that variable i stays below the value of 6. Since we start the variable at 1, it is initially smaller than 6. Then the program enters a loop and prints i, then it continues to add 1 to the current value of i. The expression:

i +=1

means, add 1 to the variable i. it could have also been written like this: i = i + 1

### Can we break the loop by force or do we have to wait for the condition to break each time?

break: Stops and exits the loop entirely. Once Python hits a break, the loop is over, no matter how many items are left to process.

continue: Skips the rest of the current iteration and jumps straight to the next item or cycle in the loop.

```python
i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1
```

```python
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
```

BTW, you can also use else as the alternate option just as in the IF statement

```python
i = 1
while i < 6:
    print(i)
    i += 1
else:
    print("i is no longer less than 6")
```
