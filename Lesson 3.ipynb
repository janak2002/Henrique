# Lesson 3: FOR loop

FOR loop is not much different than the while loop, it jsut does a slightly different thing :). As we have seen the WHILE loop goes on until a condition that ensures its running breaks. 

In a FOR loop, we ensure the running of some code for a specified number of times. In that way it differs from the WHILE loop which can virtually run forever if we forget to set the condition properly.

## 1. Looping through variables

```python
# Looping through a list

fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
```

```python
# Looping through elements of a string

for x in "banana":
    print(x)
```

```python
# Exiting the loop earlier than it was provisioned to end

fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(f"First loop: {x}")
    if x == "banana":
        break



fruits = ["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        break
    print(f"Second loop: {x}")
```

## 3. Range variable: Running for a specified NUMBER of times

If we want to repeat some code an exact amount of times (so not loop THROUGH a variable but print a result a certain NUMBER of times) we can achieve this by using a range variable

Range variable is also a variable like:

x = 3

or

x = "Jana"

but unlike the previous, it holds a whole range of numbers from 0 up until n-1.

```python
x = range(6)
print(x)
```

```python
for x in range(6):
    print(x)
```

We can use the range variable with a defined numebr of steps. Like this:

```python
for x in range(2, 30, 3):   # range(begin_index, end_index+1, steps)
    print(x)
```

```python
for x in range(6):
    if x == 3: break
    print(x)
else:
    print("Finally finished!")
```

## Nested FOR loops

Just like in the IF statements, FOR loop can be nested. Like so:

```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
    for y in fruits:
        print(x, y)
```
